# comments
```python
# This is a comment
```
Python doesn't support multi-line comments (like `/* ... */` in most other languages), but there is a workaround:

By using triple-quoted strings we can create a multi-string literal, that, if left unassigned and not used a function docstring (more on this later), will be ignored at runtime. Syntax:
```python
"""
First line.
Second line. Works (:
"""
```
or
```python
'''
First line.
Second line. Also works (:
'''
```

# variables
Syntax:
```python
name = "value"
```

First characters are restricted to letters & underscores. The rest of the name also allows numbers. Fully case sensitive:

`score` $\neq$ `Score` $\neq$ `SCORE`

Names also can't use Python reserved keywords (e.g. `if`, `else`, etc...)

## naming conventions
Standard variables:
- `snake_case`
- `camelCase`

Constants:
- `UPPER_SNAKE_CASE`

Private variables:
- `_leading_underscore`

## data types

Because Python is <b>dynamically typed</b>, you don't declare or manually assign it a data type (e.g. `int x = 5`) but you instead just do `x = 5`.

Types are thus automatically assigned at <b>runtime</b> (vs statically-typed at compile-time, e.g. Java) and there is full <b>reassignment flexibility</b>:
```python
age = 20
age = "twenty"
# this is perfectly fine, but why?
```
Because in Python <b>types belong to objects</b>, not variables. Variables in Python are just used as labels or pointers to objects in memory. The object itself knows its `type()`, while the variable has no type of its own.
